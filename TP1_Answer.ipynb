```python
# ---------------------------------------------------------------------------
# Problem 3 (Easy) · Temperature converter
# ---------------------------------------------------------------------------
def celsius_to_fahrenheit(c, round_to=1):
    """Convert degrees Celsius to degrees Fahrenheit."""
    return round(c * 9 / 5 + 32, round_to)


def fahrenheit_to_celsius(f, round_to=1):
    """Convert degrees Fahrenheit to degrees Celsius."""
    return round((f - 32) * 5 / 9, round_to)


print("\n  °C  |   °F")
print("------+-------")
for c in range(0, 101, 20):
    print(f"{c:5} | {celsius_to_fahrenheit(c):6}")

print("98.6 °F =", fahrenheit_to_celsius(98.6), "°C")
print("Round trip 37 °C:", fahrenheit_to_celsius(celsius_to_fahrenheit(37)))
```

```python
def letter_grade(score):
    if score < 0 or score > 100:
        return "Invalid"
    elif score >= 97:
        return "A+"
    elif score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
sc = [99, 90, 85.5, 72, 60, 59.9, -5, 101]
for s in sc:
    print(s, "-->", letter_grade(s))
```

```python
# ---------------------------------------------------------------------------
# Problem 6 (Medium) · Leap years and days in a month
# ---------------------------------------------------------------------------
def is_leap(year):
    """Divisible by 4, except centuries, unless divisible by 400."""
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0

def days_in_month(month, year):
    """Return the number of days in a month (1–12), or None if invalid."""
    if month < 1 or month > 12:
        return None
    if month == 2:
        return 29 if is_leap(year) else 28
    if month in (4, 6, 9, 11):
        return 30
    return 31


print()
for y in [1900, 2000, 2024, 2026]:
    print(y, "leap" if is_leap(y) else "not leap")
print("Feb 2024:", days_in_month(2, 2024), "| Feb 2026:", days_in_month(2, 2026),
      "| Apr:", days_in_month(4, 2026), "| month 13:", days_in_month(13, 2026))

leap_count = 0
for y in range(2000, 2101):
    if is_leap(y):
        leap_count += 1
print("Leap years from 2000 to 2100:", leap_count)
```

```python
# ---------------------------------------------------------------------------
# Problem 7 (Medium) · Digit tools (while loops with // and %)
# ---------------------------------------------------------------------------
def digit_sum(n):
    """Add up the digits of a whole number."""
    n = abs(n)
    total = 0 #6
    while n > 0:
        total += n % 10        # last digit
        n //= 10               # drop the last digit
    return total

def reverse_number(n):
    """Return the digits of n in reverse order, e.g. 1234 -> 4321."""
    result = 0
    while n > 0:
        result = result * 10 + n % 10
        n //= 10
    return result

def is_palindrome(n):
    """True when n reads the same in both directions."""
    return n == reverse_number(n)

print()
for n in [2026, 12321, 9875]:
    print(f"{n}: digit sum = {digit_sum(n)}, reversed = {reverse_number(n)}, "
          f"palindrome = {is_palindrome(n)}")

count = 0
for n in range(10, 1000):
    if is_palindrome(n):
        count += 1
print("Palindromes from 10 to 999:", count)
```

```python
# ---------------------------------------------------------------------------
# Problem 8 (Medium) · *args, **kwargs and lambda
# ---------------------------------------------------------------------------
def describe(*numbers, decimals=2, label="Data"):
    """Return (count, min, max, mean) of any number of values and print them."""
    if len(numbers) == 0:
        print(f"{label}: no values")
        return None
    mean = round(sum(numbers) / len(numbers), decimals)
    print(f"{label}: n={len(numbers)} min={min(numbers)} max={max(numbers)} mean={mean}")
    return len(numbers), min(numbers), max(numbers), mean

def make_tag(text, **style):
    """Wrap text in an HTML-like tag built from keyword arguments."""
    attrs = ""
    for key, value in style.items():
        attrs += f' {key}="{value}"'
    return f"<span{attrs}>{text}</span>"

print()
describe(12, 7, 30, 4)
describe(3.14159, 2.71828, decimals=3, label="Constants")
describe()
print(make_tag("Hello", color="red", size=14))

square = lambda x: x ** 2
is_adult = lambda age: age >= 18
print(square(9), is_adult(17), is_adult(21))
```

```python
def electricity_bill(kwh):

    if kwh < 0:
        return None

    tier1 = min(kwh, 50)
    tier2 = min(max(kwh - 50, 0), 150)
    tier3 = max(kwh - 200, 0)

    cost = (tier1 * 480) + (tier2 * 610) + (tier3 * 730)

    cost += 2000

    bill = cost * 1.10

    return round(bill)

consumptions = [0, 35, 50, 120, 200, 350]

print("========== ELECTRICITY BILL ==========")

for kwh in consumptions:
    bill = electricity_bill(kwh)
    usd = bill / 4100

    print(f"{kwh} kWh -> {bill:,} riel  (≈ {usd:.2f} USD)")

print("\n========== YOUR MONTHLY BILL ==========")

user_kwh = float(input("Enter your electricity consumption (kWh): "))

user_bill = electricity_bill(user_kwh)

if user_bill is None:
    print("Invalid input! Consumption cannot be negative.")
else:
    print(f"Your electricity bill: {user_bill:,} riel")
    print(f"Your bill in USD: ≈ {user_bill / 4100:.2f} USD")
```

```python
# ---------------------------------------------------------------------------
# Problem 10 (Hard) · Prime numbers and Goldbach's conjecture
# ---------------------------------------------------------------------------
def is_prime(n):
    """True when n is a prime number. Tests divisors up to the square root."""
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True


print("\nPrimes below 50:", end=" ")
count = 0
for n in range(50):
    if is_prime(n):
        print(n, end=" ")
for n in range(100):
    if is_prime(n):
        count += 1
print("\nNumber of primes below 100:", count)


def goldbach(n):
    """Return the first pair of primes (p, q) with p + q == n, or None."""
    for p in range(2, n // 2 + 1):
        if is_prime(p) and is_prime(n - p):
            return p, n - p
    return None


for n in range(4, 31, 2):
    p, q = goldbach(n)
    print(f"{n} = {p} + {q}")

# Check the conjecture for every even number from 4 to 10 000
all_ok = True
for n in range(4, 10001, 2):
    if goldbach(n) is None:
        all_ok = False
        print("Counter-example:", n)
        break
print("Goldbach holds for every even number from 4 to 10000:", all_ok)
```
